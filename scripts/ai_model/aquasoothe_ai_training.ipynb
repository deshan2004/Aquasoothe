### 🧠 Step 2: Load AudioLDM AI Model onto GPU

```python
import torch
from diffusers import AudioLDMPipeline
from IPython.display import Audio, display
import scipy.io.wavfile

print("Loading AudioLDM Diffusion Model onto T4 GPU...")
model_id = "cvssp/audioldm-s-full-v2"

# Load model in FP16 for super-fast GPU execution
pipe = AudioLDMPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda")
print("✅ AudioLDM AI Model Loaded Successfully onto GPU!")
```

### 🎵 Step 3: Generate Custom AI Soundscapes From Text Prompts

```python
prompts = [
    "gentle rain falling softly on roof with night crickets",
    "calming ocean waves washing onto sand with peaceful wind",
    "crackling campfire in a deep quiet forest at midnight"
]

for i, prompt in enumerate(prompts):
    print(f"\n🔊 [Generating {i+1}/{len(prompts)}]: '{prompt}'...")
    # Generates 5.0 seconds audio using diffusion steps
    audio = pipe(prompt, num_inference_steps=25, audio_length_in_s=5.0).audios[0]
    
    output_path = f"aquasoothe_sound_{i+1}.wav"
    scipy.io.wavfile.write(output_path, rate=16000, data=audio)
    print(f"✅ Generated: {output_path}")
    display(Audio(output_path))
```
